```python
# Identificar o catálogo e o schema atuais
spark.sql("SELECT current_catalog(), current_schema()").show(
    truncate=False
)

# Listar os catálogos disponíveis
spark.sql("SHOW CATALOGS").show(truncate=False)
```

```python
# Configuração inicial do Unity Catalog
# Projeto MBA - Novo Caged

CATALOG = "workspace"

spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {CATALOG}.bronze
    COMMENT 'Camada Bronze - Dados brutos do Novo Caged'
""")

spark.sql(f"""
    SHOW SCHEMAS IN {CATALOG}
""").show(truncate=False)
```

```python
# Configurações do projeto
catalogo = "workspace"
schema = "silver"

# Criar o schema Silver, caso ainda não exista
spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalogo}.{schema}
""")

# Confirmar criação
display(
    spark.sql(f"SHOW SCHEMAS IN {catalogo}")
)
```
